```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import geopandas as gpd

gdf_no_geom = gpd.read_file("/home/edier/GoogleDrive/INVESTIGACION/PAPERS/ELABORACION/PAPER_SAR/Data/df_catchments_kmeans.gpkg", ignore_geometry=True)

gdf_no_geom['log_lands_rec'] = np.log1p(gdf_no_geom['lands_rec'])

# Function to plot histograms with Gaussian fit and customized axis labels
def plot_histogram_with_custom_labels(data, xlabel, subplot_index):
    ax = plt.subplot(1, 2, subplot_index)
    
    # Plot histogram
    plt.hist(data, bins=20, density=True, alpha=0.6, color='g', edgecolor='black')
    
    # Fit a Gaussian distribution
    mu, std = norm.fit(data)
    
    # Plot the Gaussian fit line
    xmin, xmax = plt.xlim()
    x = np.linspace(max(0, xmin), xmax, 100)
    p = norm.pdf(x, mu, std)
    plt.plot(x, p, 'k', linewidth=2)
    
    # Set x-axis minimum to 0
    plt.xlim(left=0)
    
    # Axis label customization
    plt.xlabel(xlabel, fontsize=16.8)  # Increased by 20%
    plt.ylabel('Density', fontsize=14)
    plt.grid(True)
    
    # Customize spines for all borders
    ax.spines['top'].set_visible(True)
    ax.spines['right'].set_visible(True)
    ax.spines['bottom'].set_visible(True)
    ax.spines['left'].set_visible(True)
    
    # Set color and width for all spines
    for spine in ax.spines.values():
        spine.set_edgecolor('black')
        spine.set_linewidth(1.5)

# Plotting the customized distributions with updated axis labels
plt.figure(figsize=(10, 3))

# Landslide frequency
plot_histogram_with_custom_labels(gdf_no_geom['lands_rec'].dropna(), 'Landslide Frequency', 1)

# Logarithm of landslide frequency
plot_histogram_with_custom_labels(gdf_no_geom['log_lands_rec'].dropna(), 'Log(Landslide Frequency + 1)', 2)

# Adjust layout and save figure
plt.savefig('/home/edier/GoogleDrive/INVESTIGACION/PAPERS/ELABORACION/PAPER_SAR/Figures/figFinal/Fig2.png', dpi=500, bbox_inches='tight')
plt.show()
```
