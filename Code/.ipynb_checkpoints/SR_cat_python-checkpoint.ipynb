 # Spatial Autoregressive Models (SAR)                                                                   

Models for irregular lattice data (polygon) to predict number of recent landslides in 651 catchments located in the northern Colombian Andes.

```python
import warnings
warnings.filterwarnings('ignore')

import libpysal
from pysal.model import spreg
from pysal.lib import weights
from pysal.explore import esda
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from pandas import DataFrame
import seaborn as sbn
import researchpy as rp
import statsmodels.api as sm
import scipy.stats as stats
import statsmodels.formula.api as smf
import geopandas as gpd
import seaborn as sns
from sklearn.decomposition import PCA
from scipy.stats import boxcox
```

## Data

```python
gdf = gpd.read_file("G:/My Drive/INVESTIGACION/POSDOC/Data/Vector/df_catchments_kmeans.gpkg")
gdf.info()
```

```python
gdf["landslide_dens"]=gdf["lands_rec"]/(gdf["area"]*10e-6)
```

```python
# Apply the log transformation with adding 1 to ensure all values are positive
gdf["y_log"]= np.log(gdf['lands_rec'] + 1)
```

```python
# Perform Box-Cox transformation
gdf["y_boxcox"], lambda_value = boxcox(gdf['landslide_dens'] + 1)
```

```python
lambda_value
```

```python
from scipy.stats import rankdata
from scipy.stats import norm
# Rank-based inverse normal transformation
ranks = rankdata(gdf['landslide_dens'], method='average')
# Convert the ranks into quantiles of a standard normal distribution
gdf["y_transf"] = norm.ppf((ranks - 0.5) / len(ranks))
```

```python
pca = PCA(n_components=8).fit(dfs)
```

```python
print(pca.explained_variance_ratio_)
```

```python
plt.plot(pca.explained_variance_ratio_)
plt.xlabel('Number of components')
plt.ylabel('Explained variance')
plt.show()
```

```python
def pca_results(X_scaled, pca):

    dimensions = ['Dim. {}'.format(i) for i in range(1,len(pca.components_)+1)]

    components = pd.DataFrame(np.round(pca.components_, 4), columns = X_scaled.keys())
    components.index = dimensions

    ratios = pca.explained_variance_ratio_.reshape(len(pca.components_), 1)
    variance_ratios = pd.DataFrame(np.round(ratios, 4), columns = ['Explained Variance'])
    variance_ratios.index = dimensions

    fig, ax = plt.subplots(figsize = (14,8))

    components.plot(ax = ax, kind = 'bar')
    ax.set_ylabel("Feature Weights")
    ax.set_xticklabels(dimensions, rotation=0)

    for i, ev in enumerate(pca.explained_variance_ratio_):
        ax.text(i-0.40, ax.get_ylim()[1] + 0.05, "Expl. Var.\n %.4f"%(ev))

    return pd.concat([variance_ratios, components], axis = 1)

pca_results = pca_results(dfs, pca)
```

```python
xvector = pca.components_[0]
yvector = pca.components_[1]

xs = pca.transform(dfs)[:,0]
ys = pca.transform(dfs)[:,1]

## Para visualizar las proyecciones de cada variable en los componentes se utiliza la siguiente función
for i in range(len(xvector)):
# arrows project features (ie columns from csv) as vectors onto PC axes
    plt.arrow(0, 0, xvector[i]*max(xs), yvector[i]*max(ys),
              color='r', width=0.0005, head_width=0.0025)
    plt.text(xvector[i]*max(xs)*1.2, yvector[i]*max(ys)*1.2,
             list(dfs.columns.values)[i], color='r')

plt.tick_params('y', colors='k', labelsize=12, length=2)
plt.tick_params('x', colors='k', labelsize= 12, length=2)
plt.xlabel("Componente Principal 1", fontsize=16)
plt.ylabel("Componente Principal 2", fontsize=16)
plt.show()
```

```python
# Ensure 'knn5' and 'basin' are treated as categorical variables in the dataframe
dfs['knn5'] = dfs['knn5'].astype('category')
dfs['basin'] = dfs['basin'].astype('category')

# Fit the two models using statsmodels' automatic handling of categorical variables
m4_cluster = smf.ols("y_log ~ knn5 + area + elev_mean + rel_mean - 1", data=dfs).fit()  # Cluster model
m4_cuencas = smf.ols("y_log ~ basin + area + elev_mean + rel_mean - 1", data=dfs).fit()  # Basin model

# Map for LaTeX-friendly variable names
var_map = {
    'knn5[A]': 'Cluster A',  # Automatically generated by statsmodels
    'knn5[B]': 'Cluster B',
    'knn5[C]': 'Cluster C',
    'knn5[D]': 'Cluster D',
    'basin[Atrato]': 'Atrato',
    'basin[Cauca]': 'Cauca',
    'basin[Magdalena]': 'Magdalena',
    'area': '$A$',
    'elev_mean': '$E$',
    'rel_mean': '$H$'
}

# Create a DataFrame to store results
results_df = pd.DataFrame(columns=["Variable", "Cluster Model", "Basin Model"])

# List of variables to include in the table
variables = ['knn5[A]', 'knn5[B]', 'knn5[C]', 'knn5[D]', 
             'basin[Atrato]', 'basin[Cauca]', 'basin[Magdalena]',
             'area', 'elev_mean', 'rel_mean']

# Collect coefficients and p-values for both models
for var in variables:
    # Get the coefficient and p-value for the Cluster Model
    if var in m4_cluster.params.index:
        coef_cluster = m4_cluster.params[var]
        std_err_cluster = m4_cluster.bse[var]
        p_value_cluster = m4_cluster.pvalues[var]
        coef_cluster_str = f"{coef_cluster:.3f} ({std_err_cluster:.3f})"
        if p_value_cluster < 0.05:
            coef_cluster_str = f"\\textbf{{{coef_cluster:.3f}}} ({std_err_cluster:.3f})"
    else:
        coef_cluster_str = '-'
    
    # Get the coefficient and p-value for the Basin Model
    if var in m4_cuencas.params.index:
        coef_basin = m4_cuencas.params[var]
        std_err_basin = m4_cuencas.bse[var]
        p_value_basin = m4_cuencas.pvalues[var]
        coef_basin_str = f"{coef_basin:.3f} ({std_err_basin:.3f})"
        if p_value_basin < 0.05:
            coef_basin_str = f"\\textbf{{{coef_basin:.3f}}} ({std_err_basin:.3f})"
    else:
        coef_basin_str = '-'
    
    # Append the results to the DataFrame
    new_row = pd.DataFrame({
        "Variable": [var_map.get(var, var)],
        "Cluster Model": [coef_cluster_str],
        "Basin Model": [coef_basin_str]
    })
    results_df = pd.concat([results_df, new_row], ignore_index=True)

# Append Adjusted R-squared and AIC
adj_r2_row = pd.DataFrame({
    "Variable": ["Adj. $R^2$"],
    "Cluster Model": [f"{m4_cluster.rsquared_adj:.3f}"],
    "Basin Model": [f"{m4_cuencas.rsquared_adj:.3f}"]
})
results_df = pd.concat([results_df, adj_r2_row], ignore_index=True)

aic_row = pd.DataFrame({
    "Variable": ["AIC"],
    "Cluster Model": [f"{m4_cluster.aic:.3f}"],
    "Basin Model": [f"{m4_cuencas.aic:.3f}"]
})
results_df = pd.concat([results_df, aic_row], ignore_index=True)

# Generate LaTeX table with bold column headers and custom formatting
latex_table = results_df.to_latex(index=False, escape=False)

# Insert a midrule before "Adj. $R^2$"
latex_table = latex_table.replace("Adj. $R^2$", "\\midrule\nAdj. $R^2$")

# Add a note for p-values after the bottomrule
latex_table = latex_table.replace("\\bottomrule", "\\bottomrule\n\\multicolumn{3}{l}{\\textbf{$p<0.05$}, (standard errors)}")

# Bold the column headers
latex_table = latex_table.replace("\\toprule\nVariable", "\\toprule\n\\textbf{Variable}")
latex_table = latex_table.replace("Cluster Model", "\\textbf{Cluster Model}")
latex_table = latex_table.replace("Basin Model", "\\textbf{Basin Model}")

# Display the LaTeX table
print(latex_table)
```

```python
# Assuming y and X2 are already defined as your dependent and independent variables
# and gdf is your GeoDataFrame

# Fit the Basin model with OLS_Regimes
m6_cuenca = spreg.OLS_Regimes(y, X2,
                              gdf['cuenca'].tolist(),
                              constant_regi='many',
                              regime_err_sep=False,
                              name_y='y_transf', name_x=var2)

# Fit the Cluster model with OLS_Regimes
m6_cluster = spreg.OLS_Regimes(y, X2,
                               gdf['knn5'].tolist(),
                               constant_regi='many',
                               regime_err_sep=False,
                               name_y='y_transf', name_x=var2)

# Mapping of covariates to LaTeX-friendly variable names
var_map = {
    'constant': 'Constant',
    'area': '$A$',
    'elev_mean': '$E$',
    'rel_mean': '$H$'
}

# Variables (covariates) used in the models, include 'constant' for intercepts
variables = ['constant'] + var2  

# Define regime names for Basin and Cluster models
basin_regimes = ['Atrato', 'Cauca', 'Magdalena']
cluster_regimes = ['A', 'B', 'C', 'D']

# Initialize a list to store LaTeX rows
latex_rows = []

# Extract the number of variables per regime (this should match the number of covariates + constant)
num_vars_per_regime = len(variables)

# Iterate over regimes and extract coefficients, standard errors, and p-values for each regime
for regime_idx, (regime_basin, regime_cluster) in enumerate(zip(basin_regimes, cluster_regimes)):
    for var_idx, var in enumerate(variables):
        # Index for the flat array of betas and standard errors
        array_idx = regime_idx * num_vars_per_regime + var_idx
        
        # Basin coefficients, standard errors, and p-values
        basin_coef = m6_cuenca.betas[array_idx].item() if isinstance(m6_cuenca.betas[array_idx], np.ndarray) else m6_cuenca.betas[array_idx]
        basin_std_err = m6_cuenca.std_err[array_idx].item() if isinstance(m6_cuenca.std_err[array_idx], np.ndarray) else m6_cuenca.std_err[array_idx]
        basin_p_val = m6_cuenca.t_stat[array_idx][1]  # p-value
        
        # Cluster coefficients, standard errors, and p-values
        cluster_coef = m6_cluster.betas[array_idx].item() if isinstance(m6_cluster.betas[array_idx], np.ndarray) else m6_cluster.betas[array_idx]
        cluster_std_err = m6_cluster.std_err[array_idx].item() if isinstance(m6_cluster.std_err[array_idx], np.ndarray) else m6_cluster.std_err[array_idx]
        cluster_p_val = m6_cluster.t_stat[array_idx][1]  # p-value

        # Format the coefficient: bold if p-value < 0.05
        basin_coef_str = f"\\textbf{{{basin_coef:.3f}}}" if basin_p_val < 0.05 else f"{basin_coef:.3f}"
        cluster_coef_str = f"\\textbf{{{cluster_coef:.3f}}}" if cluster_p_val < 0.05 else f"{cluster_coef:.3f}"

        # Add the row for both the Basin and Cluster models, map variable names using `var_map`
        latex_rows.append(f"{regime_basin}-{var_map[var]} & {basin_coef_str}({basin_std_err:.2f}) & {regime_cluster}-{var_map[var]} & {cluster_coef_str}({cluster_std_err:.2f}) \\\\")

# Now handle Cluster D, adding the Cluster D coefficients and leaving the Basin column with a backslash (\)
for var_idx, var in enumerate(variables):
    array_idx = 3 * num_vars_per_regime + var_idx  # Cluster D is the 4th set of coefficients
    
    # Cluster D coefficients, standard errors, and p-values
    cluster_d_coef = m6_cluster.betas[array_idx].item() if isinstance(m6_cluster.betas[array_idx], np.ndarray) else m6_cluster.betas[array_idx]
    cluster_d_std_err = m6_cluster.std_err[array_idx].item() if isinstance(m6_cluster.std_err[array_idx], np.ndarray) else m6_cluster.std_err[array_idx]
    cluster_d_p_val = m6_cluster.t_stat[array_idx][1]  # p-value
    
    # Bold if significant
    cluster_d_coef_str = f"\\textbf{{{cluster_d_coef:.3f}}}" if cluster_d_p_val < 0.05 else f"{cluster_d_coef:.3f}"

    # Add row for Cluster D with \ for the basin part
    latex_rows.append(f"\\ & \\ & D-{var_map[var]} & {cluster_d_coef_str}({cluster_d_std_err:.2f}) \\\\")

# Add rows for Adj. R-squared and AIC
adj_r2_basin = m6_cuenca.ar2
adj_r2_cluster = m6_cluster.ar2
aic_basin = m6_cuenca.aic
aic_cluster = m6_cluster.aic

# Insert midrule before the "Adj. R^2" and "AIC"
latex_rows.append(r"\midrule")
latex_rows.append(f"Adj. R$^{{2}}$ & {adj_r2_basin:.4f} & - & {adj_r2_cluster:.4f} \\\\")
latex_rows.append(f"AIC & {aic_basin:.3f} & - & {aic_cluster:.3f} \\\\")

# Combine rows into a LaTeX table
latex_table = r"""
\begin{table}[htbp]\centering
\caption{Spatial regimes}
\label{tab:regimes}
\begin{tabular}{c c c c}
\toprule
\textbf{Basin-Variables} & \textbf{Coeff.} & \textbf{Cluster-Variables} & \textbf{Coeff.} \\ 
\midrule
""" + "\n".join(latex_rows) + r"""
\bottomrule
\multicolumn{3}{l}{\textbf{$p<0.05$}, (standard errors)}
\end{tabular}
\end{table}
"""

# Output the generated LaTeX table
print(latex_table)
```

### The K-neighbor correlogram
Further, it might be the case that spatial dependence in our mispredictions only matters for sites that are extremely close to one another, and it decays quickly with distance. To investigate this, we can examine the correlation between each site’s residual and the average of the 
th nearest neighbors’ residuals, increasing 
 until the estimate stabilizes. This main idea is central to the geostatistical concept, the correlogram, which gives the correlation between sites of an attribute being studied as distance increases.

One quick way to check whether or not what we’ve seen is unique or significant is to compare it to what happens when we just assign neighbors randomly. If what we observe is substantially different from what emerges when neighbors are random, then the structure of the neighbors embeds a structure in the residuals. We won’t spend too much time on this theory specifically, but we can quickly and efficiently compute the correlation between our observed residuals and the spatial lag of an increasingnearest neighbor set:

```python
correlations = []
nulls = []
for order in range(1, 51, 5):
    knn.reweight(k=order, inplace=True)
    knn.transform = "r"
    lag_residual = weights.spatial_lag.lag_spatial(knn, m2.u)
    random_residual = m2.u[np.random.permutation(len(m2.u))]
    random_lag_residual = weights.spatial_lag.lag_spatial(knn, random_residual)
    correlations.append(np.corrcoef(m2.u.flatten(), lag_residual.flatten())[0, 1])
    nulls.append(np.corrcoef(m2.u.flatten(), random_lag_residual.flatten())[0, 1])
```

```python
plt.rc("figure", figsize=(6,4))
plt.plot(range(1, 51, 5), correlations)
plt.plot(range(1, 51, 5), nulls, color="orangered")
plt.hlines(np.mean(correlations[-3:]),*plt.xlim(),linestyle=":",color="k")
plt.hlines(np.mean(nulls[-3:]), *plt.xlim(), linestyle=":", color="k")
plt.text(s="Long-Run Correlation: ${:.2f}$".format(np.mean(correlations[-3:])),x=25,y=0.3,)
plt.text(s="Long-Run Null: ${:.2f}$".format(np.mean(nulls[-3:])),x=25,y=0.05,)
plt.xlabel("$K$: number of nearest neighbors")
plt.ylabel("Correlation between site \n and neighborhood average of size $K$")
plt.savefig('G:/My Drive/INVESTIGACION/POSDOC/Figuras/Knn_correlogram_m2.png', dpi=300, bbox_inches='tight')
```
