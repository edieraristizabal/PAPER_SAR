```python
import matplotlib.pyplot as plt
import numpy as np

# --- Data for each variable at different bandwidths ---
variables = ["Const", "A", "E", "H"]
bandwidths = ["BW=10", "BW=50", "BW=200", "BW=500"]

# Data organized as minimum, mean, and maximum values
coefficients_min = {
    "Const": [-31.793, -1.155, 1.482, 1.697],
    "A": [-6.346, -0.026, 0.276, 0.518],
    "E": [-32.985, -2.178, -0.351, 0.336],
    "H": [-5.951, -0.816, 0.209, 0.426]
}
coefficients_mean = {
    "Const": [2.244, 1.957, 1.996, 1.908],
    "A": [0.927, 0.656, 0.616, 0.606],
    "E": [0.368, 0.404, 0.455, 0.489],
    "H": [0.526, 0.518, 0.657, 0.569]
}
coefficients_max = {
    "Const": [16.443, 5.021, 2.762, 2.063],
    "A": [12.990, 1.404, 0.780, 0.682],
    "E": [11.871, 2.828, 0.953, 0.687],
    "H": [14.452, 1.992, 1.270, 0.719]
}

# --- Calculate asymmetric error bars ---
# This is more accurate as it reflects the actual min/max range around the mean.
# The format is a list of two arrays: [lower_errors, upper_errors]
error_bars = {}
for var in variables:
    lower_error = np.array(coefficients_mean[var]) - np.array(coefficients_min[var])
    upper_error = np.array(coefficients_max[var]) - np.array(coefficients_mean[var])
    error_bars[var] = [lower_error, upper_error]

# Adjusted R² and AIC values for each bandwidth
adj_r2_values = [0.841, 0.701, 0.591, 0.523]
aic_values = [1109.968, 1410.810, 1517.876, 1585.083]

# --- Creating the plot ---
# Four subplots (1 row, 4 columns) with a shared y-axis.
fig, axes = plt.subplots(1, 4, figsize=(12, 4), sharey=True)

# Plot each bandwidth as a separate subplot
for idx, (ax, bw) in enumerate(zip(axes, bandwidths)):
    x_positions = np.arange(len(variables))
    
    # Plot each variable's mean (coefficient) with its asymmetric error bars
    for i, var in enumerate(variables):
        y_value = coefficients_mean[var][idx]
        y_error = [error_bars[var][0][idx], error_bars[var][1][idx]]
        ax.errorbar(x_positions[i], y_value, yerr=np.array(y_error).reshape(2,1), fmt='o', capsize=5, color="black", ecolor="gray")
    
    # Customize the plot with the established style
    ax.set_xticks(x_positions)
    ax.set_xticklabels(variables, rotation=90, ha="center", fontsize=12)
    ax.grid(True, linestyle="--", alpha=0.6)
    ax.axhline(0, color='red', linestyle='--', linewidth=1)
    ax.set_title(bw, fontsize=14)

    # Add Adjusted R² and AIC text in a styled box
    text_str = f"Adj. $R^2$: {adj_r2_values[idx]:.3f}\nAIC: {aic_values[idx]:.3f}"
    ax.text(0.95, 0.95, text_str, transform=ax.transAxes, fontsize=12,
            verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round,pad=0.5', fc='wheat', alpha=0.5))

# Set a single y-axis label for the entire figure
axes[0].set_ylabel("Coefficient Value", fontsize=14)

# --- Final adjustments ---
plt.tight_layout()

# Save figure
plt.savefig('/home/edier/GoogleDrive/INVESTIGACION/PAPERS/ELABORACION/PAPER_SAR/Figures/figFinal/Fig11.png', dpi=500, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Covariate names and abbreviations for x-axis labels
covariates = [
    "Mean Elevation:Cluster A", "Mean Elevation:Cluster B", "Mean Elevation:Cluster C", "Mean Elevation:Cluster D",
    "Catchment Area:Cluster A", "Catchment Area:Cluster B", "Catchment Area:Cluster C", "Catchment Area:Cluster D",
    "Mean Relief:Cluster A", "Mean Relief:Cluster B", "Mean Relief:Cluster C", "Mean Relief:Cluster D",
    "Wx.Catchment Area:Cluster A", "Wx.Catchment Area:Cluster B", "Wx.Catchment Area:Cluster C", "Wx.Catchment Area:Cluster D",
    "Wx.Mean Elevation:Cluster A", "Wx.Mean Elevation:Cluster B", "Wx.Mean Elevation:Cluster C", "Wx.Mean Elevation:Cluster D",
    "Wx.Mean Relief:Cluster A", "Wx.Mean Relief:Cluster B", "Wx.Mean Relief:Cluster C", "Wx.Mean Relief:Cluster D"
]
covariates_abbreviated = [
    "A-Elevation", "B-Elevation", "C-Elevation", "D-Elevation",
    "A-Area", "B-Area", "C-Area", "D-Area",
    "A:Relief", "B-Relief", "C-Relief", "D-Relief",
    "Wx.A-Area", "Wx.B-Area", "Wx.C-Area", "Wx.D-Area",
    "Wx.A-Elevation", "Wx.B-Elevation", "Wx.C-Elevation", "Wx.D-Elevation",
    "Wx.A-Relief", "Wx.B-Relief", "Wx.C-Relief", "Wx.D-Relief"
]

# Coefficients and standard errors for each model (SEM, SAR, SDM)
sem_values = [0.572, 0.429, 0.196, 0.673, 0.433, 0.699, 0.310, 0.715, 0.337, 0.524, -0.239, 0.738] + [None] * 12
sem_errors = [0.32, 0.17, 0.36, 0.30, 0.08, 0.06, 0.10, 0.11, 0.25, 0.12, 0.27, 0.26] + [None] * 12

sar_values = [0.357, 0.648, 0.273, 0.610, -0.125, 0.215, 0.097, 0.711, 0.080, 0.477, -0.120, 0.546] + [None] * 12
sar_errors = [0.09, 0.06, 0.10, 0.12, 0.16, 0.09, 0.19, 0.16, 0.16, 0.08, 0.23, 0.24] + [None] * 12

sdm_values = [0.354, 0.694, 0.308, 0.688, 0.851, 0.255, -0.010, 0.506, 0.451, 0.448, -0.323, 0.726,
              -0.660, -0.573, -0.376, -0.646, -1.017, -0.077, 0.081, 0.057, -0.287, -0.001, 0.280, -0.437]
sdm_errors = [0.08, 0.06, 0.10, 0.12, 0.38, 0.18, 0.41, 0.33, 0.25, 0.12, 0.28, 0.26,
              0.20, 0.15, 0.26, 0.30, 0.43, 0.22, 0.50, 0.39, 0.31, 0.15, 0.48, 0.45]

# Filtering out None values in the SEM and SAR data for plotting
def filter_none(values, errors):
    filtered_values = [v if v is not None else np.nan for v in values]
    filtered_errors = [e if e is not None else np.nan for e in errors]
    return filtered_values, filtered_errors

sem_values_filtered, sem_errors_filtered = filter_none(sem_values, sem_errors)
sar_values_filtered, sar_errors_filtered = filter_none(sar_values, sar_errors)

# Setting up plot
fig, ax = plt.subplots(figsize=(12, 8))
x = np.arange(len(covariates))  # Position for each covariate

# Plotting the coefficients with error bars
ax.errorbar(x - 0.2, sem_values_filtered, yerr=sem_errors_filtered, fmt='o', capsize=3, label='SEM + Cluster Model', color='blue')
ax.errorbar(x, sar_values_filtered, yerr=sar_errors_filtered, fmt='o', capsize=3, label='SAR + Cluster Model', color='green')
ax.errorbar(x + 0.2, sdm_values, yerr=sdm_errors, fmt='o', capsize=3, label='SDM + Cluster Model', color='red')

# Customizing the plot
ax.set_xticks(x)
ax.set_xticklabels(covariates_abbreviated, rotation=45, ha='center', fontsize=12)  # Increased font size for x-tick labels
ax.set_ylabel("Coefficient Value", fontsize=14)  # Increased font size for y-axis label
ax.legend(loc='upper right', fontsize=12)  # Increased font size for legend
ax.grid(True, linestyle="--", alpha=0.6)

# Adding border to top and right
for spine in ["top", "right", "bottom", "left"]:
    ax.spines[spine].set_linewidth(1.5)
    ax.spines[spine].set_visible(True)

# Moving R² and AIC information to the bottom left
r2_aic_text = (
    "AIC (SEM): 1523.690\n"
    "AIC (SAR): 1533.809\n"
    "AIC (SDM): 1502.588"
)
ax.text(0.02, 0.02, r2_aic_text, transform=ax.transAxes, va='bottom', fontsize=10, color='black')

# Adding lambda and rho information on the bottom right inside the plot area with colors
ax.text(0.98, 0.02, r"$\lambda$ (SEM): 0.873", transform=ax.transAxes, ha='right', va='bottom', fontsize=10, color='blue')
ax.text(0.98, 0.06, r"$\rho$ (SAR): 0.770", transform=ax.transAxes, ha='right', va='bottom', fontsize=10, color='green')
ax.text(0.98, 0.10, r"$\rho$ (SDM): 0.778", transform=ax.transAxes, ha='right', va='bottom', fontsize=10, color='red')

plt.tight_layout()
plt.savefig('/home/edier/GoogleDrive/INVESTIGACION/PAPERS/ELABORACION/PAPER_SAR/Figures/figFinal/Fig13.png', dpi=500, bbox_inches='tight')
plt.show()
```
